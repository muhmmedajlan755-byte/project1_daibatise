count the placeholders

```python
(df['glucose'] == 0).sum()
```

```python
placeholder_col=["glucose","blood_pressure","skin_thickness","insulin","bmi"]
for a in placeholder_col:
    print(a,"->",(df[a] == 0).sum(), "zeros")
```

```python
import numpy as np
```

```python
df['glucose'] = df['glucose'].replace(0,np.nan)
```

```python
before_missing = df.isnull().sum().sum()

for a in placeholder_col:
    df[a] = df[a].replace(0,np.nan)
    
after_missing = df.isnull().sum().sum()
print("missing cells before:", before_missing)
print("missing cells after:", after_missing)
```

re-measure the damage

```python
df.isnull().sum()
```

```python
df.isnull().mean()
```

fill numeric columns

```python
df['glucose'] = df['glucose'].fillna(df['glucose'].mean())
```

```python
column_fill = ['age','blood_pressure','skin_thickness','insulin','bmi']
for i in column_fill:
    df[i] = df[i].fillna(df[i].median())
```

```python
df.isnull().sum()
```

fill cateogorical columns

```python
categorical_fill=['physical_activity','smoker']
for i in categorical_fill:
     df[i] = df[i].fillna('unkown')
     
    
```
